# Diamonds Price Prediction - Exploratory Data Analysis

## 1. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load Data

```python
df = pd.read_excel(
    "diamonds.xlsx",
    sheet_name="diamonds"
)
```

## 3. Quick Data Check

```python
print("\nFirst 5 Rows:")
print(df.head())
```

```python
print("\nDataset Shape:")
print(df.shape)
```

```python
print("\nColumn Names:")
print(df.columns)
```

```python
print("\nDataset Information:")
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

## 4. Check Invalid Zero Values

```python
print("\nZero Values Check:")
print("Zero Length:", (df["Length (x)"] == 0).sum())
print("Zero Width:", (df["Width (y)"] == 0).sum())
print("Zero Height:", (df["Height (z)"] == 0).sum())
print("Zero Volume:", (df["Volume_mm3"] == 0).sum())
```

## 5. Distributions

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Price", bins=40, kde=True)
plt.title("Distribution of Diamond Prices")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Carat", bins=40, kde=True)
plt.title("Distribution of Diamond Carat")
plt.xlabel("Carat")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Volume_mm3", bins=40, kde=True)
plt.title("Distribution of Diamond Volume")
plt.xlabel("Volume (mm³)")
plt.ylabel("Frequency")
plt.show()
```

## 6. Relationships with Price

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Carat", y="Price", alpha=0.3)
plt.title("Carat vs Price")
plt.xlabel("Carat")
plt.ylabel("Price ($)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Volume_mm3", y="Price", alpha=0.3)
plt.title("Volume vs Price")
plt.xlabel("Volume (mm³)")
plt.ylabel("Price ($)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Depth", y="Price", alpha=0.3)
plt.title("Depth vs Price")
plt.xlabel("Depth")
plt.ylabel("Price ($)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Table", y="Price", alpha=0.3)
plt.title("Table vs Price")
plt.xlabel("Table")
plt.ylabel("Price ($)")
plt.show()
```

## 7. Boxplots for Outliers

## 8. Average Price by Category

### 8.1 Average Price by Cut

```python
average_price_by_cut = (
    df.groupby("Cut")["Price"]
    .mean()
    .sort_values()
)

print("\nAverage Price by Cut:")
print(average_price_by_cut)
```

```python
plt.figure(figsize=(8, 5))
average_price_by_cut.plot(kind="bar")
plt.title("Average Price by Cut")
plt.xlabel("Cut")
plt.ylabel("Average Price ($)")
plt.xticks(rotation=0)
plt.show()
```

### 8.2 Average Price by Color

```python
average_price_by_color = (
    df.groupby("Color")["Price"]
    .mean()
    .sort_values()
)

print("\nAverage Price by Color:")
print(average_price_by_color)
```

```python
plt.figure(figsize=(8, 5))
average_price_by_color.plot(kind="bar")
plt.title("Average Price by Color")
plt.xlabel("Color")
plt.ylabel("Average Price ($)")
plt.xticks(rotation=0)
plt.show()
```

### 8.3 Average Price by Clarity

```python
average_price_by_clarity = (
    df.groupby("Clarity")["Price"]
    .mean()
    .sort_values()
)

print("\nAverage Price by Clarity:")
print(average_price_by_clarity)
```

```python
plt.figure(figsize=(8, 5))
average_price_by_clarity.plot(kind="bar")
plt.title("Average Price by Clarity")
plt.xlabel("Clarity")
plt.ylabel("Average Price ($)")
plt.xticks(rotation=0)
plt.show()
```

## 9. Correlation Analysis

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

price_correlation = (
    numeric_df.corr()["Price"]
    .sort_values(ascending=False)
)

print("\nCorrelation with Price:")
print(price_correlation)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
